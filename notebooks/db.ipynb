```python
import sqlite3

import slint
```

```python
db_path = "../fixtures/Example/.eyeris/index.db"
db = sqlite3.connect(db_path)
db.row_factory = sqlite3.Row
```

```python
def dec2slint(decimal: int):
    return slint.Color(
        {
            "red": (decimal >> 16) & 0xFF,
            "green": (decimal >> 8) & 0xFF,
            "blue": decimal & 0xFF,
        }
    )

c = dec2slint(128 * 256 * 256 + 224 * 256 + 255)
c.red, c.green, c.blue
```

```python
def row2slint(item: dict) -> dict:
    item.pop("id")
    item["tags"] = item["tags"].split("|")
    item["colors"] = []
    for color_key in ["color_1", "color_2", "color_3", "color_4"]:
        if item.get(color_key):
            item["colors"].append(dec2slint(item[color_key]))
        item.pop(color_key)
    return item
```

```python
cur = db.execute("SELECT * FROM images")
rows = cur.fetchall()
images = [row2slint(dict(row)) for row in rows]
images
```
