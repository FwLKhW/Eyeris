```python
q = pl.scan_csv("../fixtures/imgs/.eyeris/index.csv")
df = q.collect()
df.describe()
```

```python
df_filled = df.with_columns([
    pl.col("tags").fill_null(""),
    pl.col("hash").fill_null(""),
])
df_filled.head(5)
```

```python
def hex_to_rgb(hex: str) -> tuple[int, int, int]:
    return (int(hex[1:3], 16),
            int(hex[3:5], 16),
            int(hex[5:7], 16))

def transform_item(item: dict) -> dict:
    item.pop("id")
    item["tags"] = item["tags"].split("|")
    item["colors"] = []
    for color_key in ["color_1", "color_2", "color_3", "color_4"]:
        if item.get(color_key):
            item["colors"].append(hex_to_rgb(item[color_key]))
        item.pop(color_key)
    return item
```

```python
items = df_filled.to_dicts()
transform_item(items[0])
```
