```python
from IPython.display import Image, display

from src.collection import Collection
```

&nbsp;

### **📂 Open collection**

```python
collection = Collection()
```

&nbsp;

### 🖼️ **Get all images**

```python
images = collection.images_df()
images.head(5)
```

&nbsp;

### ⚙️ **Indexing**

```python
collection.ss.index_images(images.to_dict(orient="records"))
```

&nbsp;

### 🔍 **Search**

```python
queries = [
    "Airplane",
    "A picture of the moon",
    "Thunder storm",
    "Flowers",
    "A picture of forest",
    "A anime cat girl in school uniform",
]

for query in queries:
    print(query)
    ids = collection.ss.search(query, thres=0.1, limit=3)
    for id in ids:
        image = images.loc[images["id"] == id]
        display(Image(filename=image["thumb_path"].item(), width=256))
```
