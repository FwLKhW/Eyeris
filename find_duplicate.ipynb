# 🔍 **Finding duplicated images**

2 images is duplicated if their pixel value exactly equal one-by-one, even when their file format or bit depth is differ.

&nbsp;

### 🤷‍♂️ *Guess what...*

*There is no duplicated image!*

That is because of...

&nbsp;

### 📐 The anti-duplicate design choice

Every images when imported to collection are placed in a folder named by its hash:

```sh
<collection>/
└── images/
    └── 1dn32hf...rbs/              # named by image hash
        └── i-am-unique.png         # image placed here
```

If `i-am-unique.png` is imported again (or any identical image), it will be refuse because we already have that exact image.

&nbsp;

### ❓ Why?

Because we don't need any IDENTICAL images. Right?
