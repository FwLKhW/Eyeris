# 📦 **Create a collection**

To run other notebooks, we need a `collection` with some indexed images.

```sh
<collection>/
├── embeddings/             # embeddings created by VLM, for semantic search
├── images/
│   ├── 1dn32hf...rbs/              # named by image hash, contain image and its thumbnail
│   │   ├── <image-name>.<image-ext>        # original image
│   │   └── thumbnail.png                   # thumbnail
│   ├── j5f49hf...asd/
│   │   └── ...
│   └── ...
└── index.db              # sqlite3 database store all images data
```

&nbsp;

````python
from pathlib import Path
from subprocess import run

from crossfiledialog import choose_folder, open_multiple
````

&nbsp;

### **📂 Open a collection**

An empty folder OR a already existed collection

````python
collection_path = Path(choose_folder())
print(f"Selected collection: {collection_path}")
````

&nbsp;

### **🖼️ Add some images**

Using Rust CLI

&nbsp;

#### Add all images in a folder 📂

````python
image_folder_path = choose_folder()
run(["target/release/cli", str(collection_path), "add", "folder", image_folder_path], check=False)
````

&nbsp;

#### or selected multiple images 🖼️

````python
image_paths = open_multiple(filter={"Images": ["*.png", "*.jpg", "*.jpeg", "*.webp"]})
run(["target/release/cli", str(collection_path), "add", "images"] + image_paths, check=False)
````
