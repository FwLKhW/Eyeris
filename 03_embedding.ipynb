### **Search**

```python
from pprint import pp

queries = [
    "Airplane",
    "A picture of the moon",
    "Thunder storm",
    "Flowers",
    "A picture of forest",
    "Leaf",
]

for query in queries:
    print(query)
    pp(ss.search(query, limit=3))
    print()
```
