### **OR** select a folder ...

```python
from pathlib import Path

from crossfiledialog import choose_folder

collection = Collection(Path(choose_folder()))
```

```python
tagger = AutoTagger()
```

### **Select ids and paths**

```python
images = collection.get_db().execute("SELECT id, path FROM images").fetchall()
image_paths = [(img["id"], collection.path / img["path"]) for img in images]
image_paths
```

### **Tag it and write**

```python
for id, image_path in image_paths:
    tags = tagger.tag(image_path)
    collection.tag(id, tags)
```
