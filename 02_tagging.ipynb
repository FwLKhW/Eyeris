```python
from src.collection import Collection
from src.config import EXAMPLE_COLLECTION_PATH
from src.tag import AutoTagger
```

```python
collection = Collection(EXAMPLE_COLLECTION_PATH)
tagger = AutoTagger()
```

### **Select ids and paths**

```python
images = collection.get_db().execute("SELECT id, path FROM images").fetchall()
image_paths = [(img["id"], EXAMPLE_COLLECTION_PATH / img["path"]) for img in images]
image_paths
```

### **Tag it and write**

```python
for id, image_path in image_paths:
    tags = tagger.tag(image_path)
    collection.tag(id, tags)
```
