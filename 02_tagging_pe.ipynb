### **📂 Open collection**

```python
collection = Collection()
```

### **🖼️ Load all images**

```python
images = collection.images_df()
images.head()
```

&nbsp;

### **🏷️ Tag it!**

Given 🏷️ **tag** -> find all 🖼️ **images** may suit with that tag

E.g. find all images may suit with tag `short_hair`

```python
tag = "short_hair"
ids = collection.ss.search(tag, thres=0.1, limit=5)
for id in ids:
    image = images.loc[images["id"] == id]
    display(Image(filename=image["thumb_path"].item(), width=256))
```

Given 🖼️ **image** + 🏷️ **tags** -> find the **tag** most suit with that image, may null

```python
tagger = PETagger()
tagger.set_tags(["river", "city", "forest", "house", "tree"])
for image in images[:5].itertuples():
    tags = tagger.tag(image.path)
    display(Image(filename=collection.path / image.thumb_path, width=256))
    print(tags)
```
