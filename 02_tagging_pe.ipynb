### **✨ Init**

### **🖼️ Load all images**

```python
images = collection.images_df()
images.head()
```

### **🏷️ Tag it!**

```python
from IPython.display import Image, display

TAGS = ["avatar", "yearbook", "motor", "japan", "other/unknown"]
tagger.set_tags(TAGS)

for image in images.itertuples(index=True):
    tags = tagger.tag(image.path)  # ty: ignore[unresolved-attribute]
    if image.Index < 10:  # ty: ignore[unresolved-attribute]
        display(Image(filename=collection.path / image.thumb_path, width=128))  # ty: ignore[unresolved-attribute]
        print(tags)
```
